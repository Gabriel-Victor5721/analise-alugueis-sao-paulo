## 3. Análise estatística dos dados

**Calculo da média e da mediana dos preços dos alugueis**

**O problema da Média:** A média é sensível a valores extremos. Se você tiver mil apartamentos comuns de R$ 2.000 e apenas uma cobertura de R\$ 100.000, essa única cobertura vai "puxar" a média de todos os outros imóveis para cima, dando a falsa impressão de que o custo de vida geral é muito mais alto do que a realidade da maioria.

**O poder da Mediana:** A mediana é o valor que fica exatamente no meio quando organizamos todos os preços do mais barato ao mais caro. Ela não se importa se existe uma cobertura de R\$ 100.000, porque ela olha apenas para o valor central da distribuição. Por isso, a mediana não é necessariamente o aluguel real de um imóvel: ela representa o centro da distribuição dos preços.

**A Conclusão:** Quando se compara a média e a mediana, a diferença entre elas mostra o quanto o mercado está distorcido por preços muito altos (ou muito baixos). Se a média for muito maior que a mediana, sabe-se que existem alguns imóveis absurdamente caros puxando o grupo para cima.

**Importante:** por enquanto, estamos usando todos os registros da coluna rent, pois os dois casos suspeitos têm preços preenchidos. Como a área não é usada nesse cálculo, não há motivo automático para excluí-los da análise de aluguel.

```python
# calculo dos valores das médias e das medianas dos preços dos alugueis
media = df["rent"].mean()
mediana = df["rent"].median()

print(f"\nValor da média dos alugueis: R$ {media:.2f}\n")
print(f"Valor da mediana dos alugueis: R$ {mediana:.2f}")
```

**Resultado**

A média dos preços anunciados de aluguel foi de R\$ 3.250,81, enquanto a mediana foi de R\$ 2.415,00. Portanto, a média ficou R\$ 835,81 acima da mediana.

Essa diferença sugere que a presença de imóveis com aluguéis mais elevados pode estar influenciando a média para cima. A mediana representa o valor central dos anúncios ordenados por preço e tende a ser menos sensível a valores extremos.

```python
display(df["rent"].describe().round(2))
```

**describe():** O método df.describe() da biblioteca Pandas em Python gera um resumo estatístico completo das colunas de um DataFrame

  ficaremos com o min e o max que calcula respectivamente o menor valor e o maior valor registrado, com isso decobre-se que o maior valor já registrado dos alugueis é de R\$ 25000,00.O valor máximo pode contribuir para elevar a média, mas não podemos afirmar que ele é o responsável por toda a diferença entre a média e a mediana sem investigar a distribuição dos preços. Outros aluguéis elevados também podem influenciar esse resultado.

## 4. Análise de relações entre variáveis



**Correlação entre a área do imóvel e o preço do aluguel**

Nesta etapa, foi realizada uma análise para investigar a relação entre a área dos imóveis e seus respectivos preços de aluguel.

Para isso, foi utilizado o coeficiente de correlação de Pearson, uma medida estatística que permite identificar a direção e a intensidade da relação linear entre duas variáveis numéricas.

Antes do cálculo, foram selecionados apenas os registros cuja área é maior que zero, pois foram identificados dois imóveis com área igual a zero na etapa de verificação da qualidade dos dados.

Dessa forma, o cálculo considera apenas os registros com valores de área válidos para essa análise.

```python
df_area_valida = df[df["area"] > 0]
print(df_area_valida[["area", "rent"]].corr())
```

**Resultado:**

Para investigar a relação entre o tamanho dos imóveis e seus respectivos preços de aluguel, foi calculado o coeficiente de correlação entre as colunas `area` e `rent`.

O resultado foi aproximadamente 0,67, indicando uma correlação positiva. Isso significa que, de maneira geral, imóveis com áreas maiores tendem a apresentar valores de aluguel mais elevados.

Entretanto, essa relação não é perfeita, pois o preço do aluguel também pode ser influenciado por outros fatores, como localização, quantidade de quartos e características do imóvel.

Para esta análise, foram considerados apenas os registros com área superior a zero, evitando que os dois registros com área igual a zero interferissem no resultado.

**ANÁLISE DOS PREÇO POR QUANTIDADE DE QUARTOS**

```python
resumo_quartos = (
    df.groupby("bedrooms")
    .agg(
        quantidade=("rent", "count"),
        aluguel_medio=("rent", "mean")
    )
    .sort_index()
)

display(resumo_quartos.round(2))
```

**Relação entre a quantidade de quartos e o preço do aluguel**

Nesta etapa, foi analisada a relação entre a quantidade de quartos dos imóveis e seus respectivos preços de aluguel.

Os resultados indicam uma tendência de aumento no preço médio conforme cresce a quantidade de quartos. Os imóveis com 1 quarto apresentaram média de R\$ 2.153,96, enquanto aqueles com 3 quartos apresentaram média de R\$ 4.795,24. Já os imóveis com 5 quartos registraram a maior média, de R\$ 8.568,25.

Entretanto, essa tendência não foi totalmente crescente, pois os imóveis com 6 quartos apresentaram média de R\$ 8.346,30, ligeiramente inferior à categoria de 5 quartos.

É importante considerar a quantidade de anúncios em cada categoria. Os imóveis com 0 e 6 quartos possuem poucas observações, o que exige cautela na interpretação dessas médias.

De maneira geral, os resultados sugerem uma relação entre a quantidade de quartos e o preço anunciado do aluguel, embora outros fatores também possam influenciar os valores.

## Análise dos preços de aluguel por distrito

Nesta etapa, será realizada uma análise dos preços de aluguel agrupando por distrito da cidade de São Paulo.

Primeiramente, será contabilizada a quantidade de anúncios disponíveis em cada distrito. Essa etapa permite identificar quais localidades possuem maior representação no conjunto de dados.

Em seguida, será calculada a média dos preços de aluguel por distrito, considerando apenas aqueles que possuem pelo menos 50 registros. Esse critério foi adotado para reduzir a influência de localidades com poucas observações, que poderiam apresentar médias pouco representativas.

Por fim, os distritos serão organizados de acordo com o valor médio do aluguel, permitindo identificar as localidades com os maiores e menores preços anunciados.

```python

# Contando quantos anúncios existem em cada distrito
anuncios_por_distrito = (
    df.groupby("district")
      .size()
      .sort_values(ascending=False)
)

display(anuncios_por_distrito.head(10))
```

## Calculo da média de alugueis por distrito:

**10 DISTRITOS COM MAIORES MÉDIAS:**



```python
resumo_distritos = (
    df.groupby("district")
    .agg(
        quantidade=("rent", "count"),
        aluguel_medio=("rent", "mean")
    )
)

resumo_distritos = resumo_distritos[
    resumo_distritos["quantidade"] >= 50
]

resumo_distritos = resumo_distritos.sort_values(
    by="aluguel_medio",
    ascending=False
)

display(resumo_distritos.head(10).round(2))
```

**10 DISTRITOS COM MENORES MÉDIAS**

```python
display(resumo_distritos.tail(10).round(2))
```

**RESULTADO DA ANÁLISE POR DISTRITOS**

Após agrupar os anúncios por distrito e considerar apenas localidades com pelo menos 50 registros, foram identificadas diferenças significativas nos preços médios de aluguel.

Cerqueira César apresentou a maior média, com R\$ 6.750,66, considerando 140 anúncios. Em seguida, aparecem Jardim Paulista, com R\$ 6.305,96, e Paraíso, com R\$ 6.285,56.

Entre os distritos com menores médias, destacam-se Sé, com R\$ 1.593,30, Campos Elíseos, com R\$ 1.646,79, e Liberdade, com R$\ 1.831,74.

A diferença entre a maior e a menor média foi de R\$ 5.157,36. Esses resultados indicam que os preços anunciados de aluguel variam consideravelmente entre os distritos analisados.

É importante considerar que essa comparação representa os anúncios disponíveis no conjunto de dados e não necessariamente todos os imóveis existentes nessas localidades.

**COMPARAÇÃO DOS PREÇOS MÉDIOS ENTRE DISTRITOS:**
